**Data Quality Analysis Summary**

```python
summary = pd.DataFrame({
    "Table": ["Customers", "Restaurants", "Menu Items", "Orders", "Payments"],
    "Missing Values": [
        customers.isnull().sum().sum(),
        restaurants.isnull().sum().sum(),
        menu_items.isnull().sum().sum(),
        orders.isnull().sum().sum(),
        payments.isnull().sum().sum()
    ],
    "Duplicate Rows": [
        customers.duplicated().sum(),
        restaurants.duplicated().sum(),
        menu_items.duplicated().sum(),
        orders.duplicated().sum(),
        payments.duplicated().sum()
    ]
})

summary["Action"] = summary.apply(
    lambda row: "Clean" if row["Missing Values"] > 0 or row["Duplicate Rows"] > 0 else "Good",
    axis=1
)
summary
```

**Clean Customer Dataset**

```python
customers_clean = customers.copy()

# Ensure customer_id is integer
customers_clean["customer_id"] = pd.to_numeric(customers_clean["customer_id"], errors="coerce")
customers_clean = customers_clean[customers_clean["customer_id"].notnull()]
customers_clean["customer_id"] = customers_clean["customer_id"].astype(int)

# Convert signup_date to datetime
customers_clean["signup_date"] = pd.to_datetime(customers_clean["signup_date"], dayfirst=True, errors="coerce")

# Handle missing values in customer_segment (fill missing with 'Unknown' or mode)
customers_clean["customer_segment"] = customers_clean["customer_segment"].fillna("Unknown")

# Strip leading/trailing whitespaces in categorical text
customers_clean["customer_name"] = customers_clean["customer_name"].str.strip()
customers_clean["city"] = customers_clean["city"].str.strip()
customers_clean["customer_segment"] = customers_clean["customer_segment"].str.strip()

# Verification check
print("Cleaned Customers Shape:", customers_clean.shape)
print(customers_clean.isnull().sum())
```

**Clean Restaurant & Menu Items Datasets**

```python
restaurants_clean = restaurants.copy()

# Ensure primary key is integer
restaurants_clean["restaurant_id"] = restaurants_clean["restaurant_id"].astype(int)

# Text standardization
restaurants_clean["restaurant_name"] = restaurants_clean["restaurant_name"].str.strip()
restaurants_clean["city"] = restaurants_clean["city"].str.strip()
restaurants_clean["cuisine"] = restaurants_clean["cuisine"].str.strip()

# Verify rating range (1.0 to 5.0)
print("Restaurant Rating Summary:")
print(restaurants_clean["rating"].describe())
```

```python
menu_items_clean = menu_items.copy()

# Ensure keys and types
menu_items_clean["item_id"] = menu_items_clean["item_id"].astype(int)
menu_items_clean["restaurant_id"] = menu_items_clean["restaurant_id"].astype(int)
menu_items_clean["item_name"] = menu_items_clean["item_name"].str.strip()
menu_items_clean["category"] = menu_items_clean["category"].str.strip()

# Verify price validity (no zero or negative prices)
print("Menu Items Price Summary:")
print(menu_items_clean["price"].describe())
print("\nItems with price <= 0:", (menu_items_clean["price"] <= 0).sum())
```

**Clean Orders Dataset**

```python
orders_clean = orders.copy()

# Convert order_date to datetime format
orders_clean["order_date"] = pd.to_datetime(orders_clean["order_date"], dayfirst=True, errors="coerce")

# Ensure numeric datatypes
orders_clean["order_id"] = orders_clean["order_id"].astype(int)
orders_clean["customer_id"] = orders_clean["customer_id"].astype(int)
orders_clean["restaurant_id"] = orders_clean["restaurant_id"].astype(int)

# Standardize string fields
orders_clean["order_status"] = orders_clean["order_status"].str.strip()

# Handle missing delivery_minutes
# Notice missing delivery_minutes belong to Cancelled orders: impute with 0 or a sentinel value
orders_clean["delivery_minutes"] = orders_clean["delivery_minutes"].fillna(0)

# Check order_amount and discount validity
print("Orders Summary:")
print(orders_clean[["delivery_minutes", "order_amount", "discount"]].describe())
print("\nOrders where discount exceeds order_amount:", (orders_clean["discount"] > orders_clean["order_amount"]).sum())
```

**Clean Payments Dataset**

```python
payments_clean = payments.copy()

# Convert IDs and Dates
payments_clean["payment_id"] = payments_clean["payment_id"].astype(int)
payments_clean["order_id"] = payments_clean["order_id"].astype(int)
payments_clean["payment_date"] = pd.to_datetime(payments_clean["payment_date"], errors="coerce")

# Clean text categories
payments_clean["payment_method"] = payments_clean["payment_method"].str.strip()
payments_clean["payment_status"] = payments_clean["payment_status"].str.strip()

print("Cleaned Payments Shape:", payments_clean.shape)
print(payments_clean.isnull().sum())
```

**Validate Relational Integrity (Foreign Keys)**

```python
# Check 1: Do all customer_ids in orders exist in customers?
invalid_customers = orders_clean[~orders_clean["customer_id"].isin(customers_clean["customer_id"])]
print("Orders with invalid customer_id:", len(invalid_customers))

# Check 2: Do all restaurant_ids in orders exist in restaurants?
invalid_restaurants = orders_clean[~orders_clean["restaurant_id"].isin(restaurants_clean["restaurant_id"])]
print("Orders with invalid restaurant_id:", len(invalid_restaurants))

# Check 3: Do all order_ids in payments exist in orders?
invalid_payments = payments_clean[~payments_clean["order_id"].isin(orders_clean["order_id"])]
print("Payments with invalid order_id:", len(invalid_payments))

# Check 4: Do all restaurant_ids in menu_items exist in restaurants?
invalid_menu_restaurants = menu_items_clean[~menu_items_clean["restaurant_id"].isin(restaurants_clean["restaurant_id"])]
print("Menu items with invalid restaurant_id:", len(invalid_menu_restaurants))
```

**Export Cleaned Datasets**

**Save Cleaned Datasets into CSV Files**

```python
customers_clean.to_csv("Cleaned_Customers.csv", index=False)
restaurants_clean.to_csv("Cleaned_Restaurants.csv", index=False)
menu_items_clean.to_csv("Cleaned_Menu_Items.csv", index=False)
orders_clean.to_csv("Cleaned_Orders.csv", index=False)
payments_clean.to_csv("Cleaned_Payments.csv", index=False)

print("All cleaned datasets successfully exported to CSV!")
```
